## Architecture

The system includes three services:

- **facade-service**: Handles POST/GET HTTP requests from the client
- **logging-service**: Stores messages with deduplication and retry support
- **messages-service**: Stores messages via queue and is able to return

## Running the Services

Run in separate terminals:

```bash
./start_services.sh
```
