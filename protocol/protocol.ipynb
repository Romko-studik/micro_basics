## Architecture

The system includes three services:

- **facade-service**: Handles POST/GET HTTP requests from the client
- **logging-service**: Stores messages with deduplication and retry support
- **messages-service**: Returns a static message over HTTP

## Running the Services

Run in separate terminals:

```bash
# Terminal 1
cd logging-service
uvicorn main:app --port 5000

# Terminal 2
cd facade-service
uvicorn main:app --port 6000

# Terminal 3
cd messages-service
uvicorn main:app --port 7000
```
